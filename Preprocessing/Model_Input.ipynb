# PHASE 4: FEATURE ENGINEERING 

## 1. Generating Numeric Time

```python
import pandas as pd
import numpy as np

df = pd.read_parquet(r"E:\INCOIS_Internhip\Data\clean_data.parquet")

print(df["DATES"].head())
print( f'Reference day = {df["DATES"].min()}')
ref_date = df["DATES"].min()
df["time_days"] = (df["DATES"] - ref_date).dt.total_seconds() / 86400
print(df["time_days"].head())

#Checks
print(df["time_days"].describe())
```

## 2. Canonical Profile ID

```python
df['profile_id'] = df.groupby(['WMOID', 'CYCLE']).ngroup()
print("Number of profiles:", df['profile_id'].nunique())
```

## 3. Sorting Depth within each profile

```python
df = df.sort_values(['profile_id', 'Depth(m)']).reset_index(drop=True)
```

## 4. Raw Feature Set

```python
BASE_FEATURES = [
    'TEMP(degC)',
    'Depth(m)',
    'LAT',
    'LON',
    'time_days',
    'doy_sin',
    'doy_cos'
]
```

## 5. Saving Raw Data

```python
df.to_parquet("Raw_data.parquet")
```
